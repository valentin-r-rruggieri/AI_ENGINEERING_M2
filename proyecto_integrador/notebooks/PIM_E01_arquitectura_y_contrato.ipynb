## Problema

Un chatbot de RR.HH. debe contestar consultas repetidas sin inventar políticas ni perder la fuente que respalda cada respuesta.

## Conceptos clave

El sistema separa pipeline de datos y pipeline de consulta. La frontera pública es una respuesta con exactamente tres claves: pregunta, respuesta y chunks relacionados.

```python
# Ejemplo ejecutable: Laboratorio 1 — Arquitectura y contrato del chatbot FAQ
import matplotlib.pyplot as plt
steps=["FAQ","chunks","índice","Top-K","prompt","JSON"]
plt.figure(figsize=(9,2));plt.plot(range(len(steps)),[0]*len(steps),"o-",color="#1f4e79")
for i,s in enumerate(steps):plt.text(i,.1,s,ha="center")
plt.axis("off");plt.title("Arquitectura mínima del FAQ RAG");plt.show()
```
