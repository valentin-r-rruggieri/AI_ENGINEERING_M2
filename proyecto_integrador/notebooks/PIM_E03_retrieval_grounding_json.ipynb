## Problema

La recuperación debe elegir de dos a cinco chunks relevantes antes de llamar al modelo. Si no hay evidencia, la respuesta debe abstenerse.

## Conceptos clave

Top-K entrega candidatos; el contexto los vuelve auditables; la generación queda acotada a lo que dicen las fuentes.

```python
# Ejemplo ejecutable: Laboratorio 3 — Retrieval, grounding y JSON
import matplotlib.pyplot as plt
stages=["query","embedding","Top-K","contexto","LLM","JSON"]
plt.figure(figsize=(9,2));plt.plot(range(6),[0]*6,"o-",color="#2a9d8f")
for i,s in enumerate(stages):plt.text(i,.1,s,ha="center")
plt.axis("off");plt.title("Consulta: retrieval antes de generation");plt.show()
```

```python
# Ejemplo ejecutable: Laboratorio 3 — Retrieval, grounding y JSON
chunks=[{"chunk_id":"faq-003","source":"faq_document.txt","score":.91,"content":"El enlace de contraseña vence en 30 minutos."},{"chunk_id":"faq-004","source":"faq_document.txt","score":.72,"content":"Soporte verifica el correo laboral."}]
context="\n\n".join(f"[{c['chunk_id']} | {c['score']}] {c['content']}" for c in chunks)
print(context)
print("Regla: sin evidencia suficiente → 'No hay información suficiente en el contexto.'")
```
