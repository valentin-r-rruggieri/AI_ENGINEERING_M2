## Modelo mental

El prompt delimita contexto, pregunta y política de abstención. Grounding es una propiedad del flujo completo, no solo una frase en el prompt.

```python
import matplotlib.pyplot as plt
steps=["Top-K","contexto\ncon IDs","prompt\ngrounded","respuesta"]
plt.figure(figsize=(8,2));plt.plot(range(4),[0]*4,"o-",color="#1f4e79")
for i,s in enumerate(steps):plt.text(i,.1,s,ha="center")
plt.axis("off");plt.title("La evidencia llega antes que la generación");plt.show()
```
