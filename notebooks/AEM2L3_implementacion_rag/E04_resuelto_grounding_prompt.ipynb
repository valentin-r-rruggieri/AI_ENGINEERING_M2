# Prompt con grounding

## Problema real

El modelo no debe completar con conocimiento externo

## Conceptos clave

El prompt delimita contexto, pregunta y política de abstención. Grounding es una propiedad del flujo completo, no solo una frase en el prompt.

```python
# Ejemplo ejecutable: Prompt con grounding
import matplotlib.pyplot as plt
steps=["Top-K","contexto\ncon IDs","prompt\ngrounded","respuesta"]
plt.figure(figsize=(8,2));plt.plot(range(4),[0]*4,"o-",color="#1f4e79")
for i,s in enumerate(steps):plt.text(i,.1,s,ha="center")
plt.axis("off");plt.title("La evidencia llega antes que la generación");plt.show()
```

## Ejemplo ejecutable

El bloque siguiente ilustra una implementación mínima del concepto.


```python
# Ejemplo ejecutable: Prompt con grounding
context="[faq-001] El enlace vence en 30 minutos."
prompt=f"Usá solo este contexto. Si falta evidencia, abstenerse.\n{context}\nPregunta: ¿cuándo vence?"
print(prompt)
```
