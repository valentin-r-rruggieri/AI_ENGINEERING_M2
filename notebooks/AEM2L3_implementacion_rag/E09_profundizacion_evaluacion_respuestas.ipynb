# Evaluar respuestas RAG

## Problema real

Separar retrieval, grounding y redacción

## Conceptos clave

Un sistema puede recuperar bien y contestar mal, o recuperar mal y redactar una respuesta convincente. La evaluación debe distinguir ambos fallos.

```python
# Ejemplo ejecutable: Evaluar respuestas RAG
import matplotlib.pyplot as plt
metrics={"retrieval":.86,"grounding":.74,"claridad":.91}
plt.bar(metrics.keys(),metrics.values(),color="#457b9d");plt.ylim(0,1);plt.title("Rúbrica: dimensiones separadas");plt.show()
```
