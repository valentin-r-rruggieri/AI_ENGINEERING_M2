## Modelo mental

El contexto no es una lista opaca: cada bloque debe conservar chunk_id, fuente y score para permitir auditoría posterior.

```python
import matplotlib.pyplot as plt
blocks=["[faq-003]\nfuente + score","[faq-008]\nfuente + score","pregunta"]
plt.figure(figsize=(8,2))
for i,b in enumerate(blocks):plt.text(i,0,b,ha="center",bbox={"boxstyle":"round","fc":"#dbe9f6"})
plt.xlim(-.7,2.7);plt.ylim(-1,1);plt.axis("off");plt.title("Contexto ensamblado y trazable");plt.show()
```
