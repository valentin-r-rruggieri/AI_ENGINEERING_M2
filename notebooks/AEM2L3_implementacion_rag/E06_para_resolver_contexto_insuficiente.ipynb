## Conceptos clave

Top-K siempre puede devolver chunks, pero esos chunks pueden ser irrelevantes. El sistema debe diferenciar 'hay resultados' de 'hay evidencia'.

```python
# Ejemplo ejecutable: Contexto insuficiente y abstención
import matplotlib.pyplot as plt
scores=[.91,.67,.42,.18]
plt.bar(range(4),scores,color=["#2a9d8f","#2a9d8f","#e9c46a","#e76f51"]);plt.axhline(.5,color="black",linestyle="--")
plt.title("Top-K no reemplaza un criterio de suficiencia");plt.xlabel("candidatos");plt.ylabel("score");plt.show()
```

## Ejemplo ejecutable

El bloque siguiente ilustra una implementación mínima del concepto.


```python
# Ejemplo ejecutable: Contexto insuficiente y abstención
results=[{"id":"faq-1","score":.84},{"id":"faq-2","score":.31}]
sufficient=[r for r in results if r["score"]>=.5]
print("abstener" if not sufficient else sufficient)
```
