# Carga y normalización

## Problema real

La calidad empieza antes del embedding

## Conceptos clave

Cargar, limpiar y validar protege al retrieval de páginas vacías, encabezados repetidos y codificación defectuosa.

```python
# Ejemplo ejecutable: Carga y normalización
import re, matplotlib.pyplot as plt
raw="  Encabezado repetido\n\nLa política de vacaciones requiere aprobación.   "
clean=re.sub(r"\s+"," ",raw).strip()
plt.figure(figsize=(8,2));plt.text(.02,.65,"antes: "+raw.replace("\n"," ↵ "),wrap=True);plt.text(.02,.2,"después: "+clean,wrap=True);plt.axis("off");plt.title("Normalización conserva significado");plt.show()
```

## Ejemplo ejecutable

El bloque siguiente ilustra una implementación mínima del concepto.


```python
# Ejemplo ejecutable: Carga y normalización
def clean(text): return re.sub(r"\s+"," ",text).strip()
print(clean("  una\n política   clara "))
```
