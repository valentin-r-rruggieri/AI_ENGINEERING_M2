# Reranking y caché

## Problema real

Precisión y latencia en etapas separadas

## Conceptos clave

El embedding recupera candidatos baratos; un reranker reordena pocos resultados. La caché evita trabajo repetido, pero debe invalidarse por versión de corpus.

```python
# Ejemplo ejecutable: Reranking y caché
import matplotlib.pyplot as plt
stages=["embedding","Top-20","reranker","Top-5","cache"]
plt.figure(figsize=(8,2));plt.plot(range(5),[0]*5,"o-",color="#6a4c93")
for i,s in enumerate(stages):plt.text(i,.1,s,ha="center")
plt.axis("off");plt.title("Precisión y latencia por etapas");plt.show()
```

## Ejemplo ejecutable
