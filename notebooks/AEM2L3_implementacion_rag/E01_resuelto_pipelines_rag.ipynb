## Modelo mental

La ingesta transforma fuentes en un índice versionado; la inferencia transforma una pregunta en evidencia y respuesta. Mezclarlos impide depurar cambios.

```python
import matplotlib.pyplot as plt
lanes={"ingesta":["fuente","limpiar","chunking","embeddings","índice"],"consulta":["pregunta","embedding","Top-K","contexto","respuesta"]}
fig,ax=plt.subplots(figsize=(9,3))
for y,(lane,steps) in enumerate(lanes.items()):
    for x,step in enumerate(steps):
        ax.text(x,y,step,ha="center",va="center",bbox={"boxstyle":"round","fc":"#dbe9f6"})
        if x: ax.annotate("",(x-.22,y),(x-.72,y),arrowprops={"arrowstyle":"->"})
    ax.text(-1,y,lane,weight="bold")
ax.set(xlim=(-1.5,4.5),ylim=(-.6,1.6),yticks=[],xticks=[],title="Dos pipelines conectados");plt.show()
```
