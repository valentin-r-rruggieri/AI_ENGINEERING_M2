# Contrato JSON con Pydantic

## Problema real

La salida debe ser validable por otro sistema

## Conceptos clave

Una respuesta RAG útil para producción contiene texto y procedencia bajo un esquema fijo. La validación rechaza claves faltantes o inventadas.

```python
# Ejemplo ejecutable: Contrato JSON con Pydantic
import matplotlib.pyplot as plt
keys=["user_question","system_answer","chunks_related"]
plt.figure(figsize=(7,2.5))
for i,key in enumerate(keys):plt.text(i,0,key,ha="center",bbox={"boxstyle":"round","fc":"#fce5cd"})
plt.xlim(-.7,2.7);plt.ylim(-1,1);plt.axis("off");plt.title("Contrato público exacto");plt.show()
```

## Ejemplo ejecutable

El bloque siguiente ilustra una implementación mínima del concepto.


```python
# Ejemplo ejecutable: Contrato JSON con Pydantic
from pydantic import BaseModel
class Answer(BaseModel):
    user_question:str; system_answer:str; chunks_related:list[dict]
print(Answer(user_question="¿Cómo?",system_answer="Con evidencia.",chunks_related=[{"chunk_id":"faq-1"}]).model_dump())
```

## Consideraciones técnicas

Confiar en que un LLM siempre respeta el formato; aceptar claves extra.
