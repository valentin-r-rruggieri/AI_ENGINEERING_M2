# Chroma persistente

## Problema real

El índice necesita memoria entre ejecuciones

## Conceptos clave

Un vector store persistente guarda documentos, IDs, embeddings y metadata para que la consulta no dependa de reconstruir el corpus.

```python
# Ejemplo ejecutable: Chroma persistente
import matplotlib.pyplot as plt
records=[("faq-001","contraseña"),("faq-002","vacaciones"),("faq-003","licencias")]
plt.figure(figsize=(7,3))
for i,(identifier,label) in enumerate(records):
    plt.text(i,0,f"{identifier}\n{label}",ha="center",bbox={"boxstyle":"round","fc":"#d9ead3"})
plt.xlim(-.7,2.7);plt.ylim(-1,1);plt.axis("off");plt.title("Colección persistente: ID + documento + metadata");plt.show()
```

## Ejemplo ejecutable

El bloque siguiente ilustra una implementación mínima del concepto.


```python
# Ejemplo ejecutable: Chroma persistente
collection={"ids":["faq-001"],"documents":["El enlace vence."],"metadatas":[{"source":"faq.txt"}]}
print(collection)
```
