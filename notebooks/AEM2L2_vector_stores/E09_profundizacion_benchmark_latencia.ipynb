# Benchmark de latencia

## Problema real

p50 y p95, no solo promedios

## Conceptos clave

El benchmark repite consultas sobre el mismo corpus, registra configuración y reporta calidad junto con percentiles de latencia.

```python
# Ejemplo ejecutable: Benchmark de latencia
latency=[3,4,3,5,4,4,5,18,4,3]
print({"p50":float(np.percentile(latency,50)),"p95":float(np.percentile(latency,95))})
```
