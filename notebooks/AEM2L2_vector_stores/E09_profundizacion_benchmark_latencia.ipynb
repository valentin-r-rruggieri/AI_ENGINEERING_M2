# Benchmark de latencia

## Problema real

El promedio esconde a los usuarios que peor la pasan

## Conceptos clave

p50 es la mediana: la mitad de las preguntas a SoporteBot responden en ese tiempo o menos. p95 es la cola lenta: el 5% de las preguntas más lentas tardan al menos eso. Un promedio bajo puede esconder un p95 alto -y esos son justo los usuarios que van a decir que "el chatbot es lento".

```python
import numpy as np

# Tiempos de respuesta (en ms) de 10 preguntas reales a SoporteBot.
latencias_ms = [3, 4, 3, 5, 4, 4, 5, 18, 4, 3]

p50 = np.percentile(latencias_ms, 50)
p95 = np.percentile(latencias_ms, 95)
print({"p50_ms": float(p50), "p95_ms": float(p95)})
```

## Por qué p95 y no solo el promedio

El promedio de esta lista es ~5.3ms, un número tranquilizador. Pero hay una consulta que tardó 18ms: si eso le pasa a cada usuario nuevo que entra al sistema, el promedio no te lo va a mostrar. El p95 sí.

## Errores comunes

Medir la latencia con una sola pregunta ("probé y anduvo rápido") y no anotar con qué versión del índice, qué modelo de embeddings ni qué tamaño de corpus se hizo la medición.

## Dónde vive esa cola lenta

El histograma siguiente muestra la distribución completa de latencias: fijate dónde queda esa consulta de 18ms, lejos del resto.

```python
import numpy as np
import matplotlib.pyplot as plt

latencias_ms = np.array([3, 4, 3, 5, 4, 4, 5, 18, 4, 3])
plt.hist(latencias_ms, bins=6, color="#457b9d")
plt.title("Distribución de latencia de SoporteBot")
plt.xlabel("ms")
plt.show()
```
