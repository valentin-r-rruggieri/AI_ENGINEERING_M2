# Comparar índices con las mismas reglas

## Problema real

No podés decidir si FAISS o Chroma es mejor si les diste inputs distintos

## Conceptos clave

Para que una comparación sea justa, los dos backends (por ejemplo FAISS y Chroma) tienen que recibir los mismos chunks, el mismo modelo de embeddings, las mismas preguntas y el mismo K. Recién ahí podés comparar Recall@K y latencia entre ellos.

```python
# Mismos chunks, mismas preguntas, mismo K=2: comparamos FAISS contra Chroma.
resultados = [
    {"backend": "faiss", "top_ids": ["faq-1", "faq-2"], "recall_at_2": 1.0, "p95_ms": 4},
    {"backend": "chroma", "top_ids": ["faq-1", "faq-2"], "recall_at_2": 1.0, "p95_ms": 6},
]

for fila in resultados:
    print(fila)
```

## Cómo leer esta comparación

Los dos backends encontraron los mismos 2 chunks (`recall_at_2` = 1.0 para ambos), así que en calidad empatan. La diferencia está en la latencia: FAISS respondió en 4ms, Chroma en 6ms. Con eso -y no antes- podés justificar por qué elegís uno u otro para SoporteBot.

## Errores comunes

Comparar backends con corpus distintos, con un K distinto, o sin un "golden set" de preguntas con respuesta conocida de antemano: cualquier diferencia que veas después no vas a poder explicar si es real o es ruido.

## Recall y latencia, lado a lado

El siguiente gráfico compara los tres puntos de referencia: búsqueda exacta (la vara de medir), FAISS y Chroma.

```python
import matplotlib.pyplot as plt

backends = ["exacto", "FAISS", "Chroma"]
recall = [1, .91, .91]
p95 = [18, 4, 6]

fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(backends, recall, color="#2a9d8f")
ax.set_ylim(0, 1)
ax.set_ylabel("Recall@5")

ax2 = ax.twinx()
ax2.plot(backends, p95, "o-", color="peru")
ax2.set_ylabel("p95 (ms)")

plt.title("Comparación bajo el mismo golden set de preguntas")
plt.show()
```
