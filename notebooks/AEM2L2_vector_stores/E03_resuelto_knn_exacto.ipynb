# k-NN exacto como baseline

## Problema real

Comparar contra todos antes de aproximar

## Conceptos clave

La búsqueda exacta compara la consulta con cada vector. Es más lenta a escala, pero ofrece la referencia para medir recall de ANN.

```python
# Ejemplo ejecutable: k-NN exacto como baseline
import numpy as np
scores=np.array([.34,.91,.72,.18])
print("Top-3 exacto:",np.argsort(scores)[::-1][:3].tolist())
```
