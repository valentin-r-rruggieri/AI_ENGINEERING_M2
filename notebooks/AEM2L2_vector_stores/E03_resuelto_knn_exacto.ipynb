# k-NN exacto como punto de partida

## Problema real

Antes de aproximar, hay que saber cuál es la respuesta correcta

## Conceptos clave

La búsqueda exacta (k-NN) compara la pregunta contra **todos** los embeddings del vector store y devuelve los k más parecidos. Es la referencia contra la que después vas a medir si un índice aproximado (ANN) está devolviendo buenos resultados.

```python
# SoporteBot compara la pregunta del usuario contra 4 FAQs, una por una.
import numpy as np

faqs = ["wifi", "facturacion", "contrasena", "horarios"]
scores = np.array([.34, .91, .72, .18])  # similitud de cada FAQ con la pregunta

top_3 = np.argsort(scores)[::-1][:3]
print("Top-3 exacto:", [faqs[i] for i in top_3])
```

## Por qué esto es el "baseline"

Con 4 FAQs, comparar contra todas cuesta nada. El problema aparece cuando el vector store tiene cientos de miles de chunks: ahí un índice ANN te va a devolver una respuesta en milisegundos, pero necesitás el resultado de k-NN exacto para poder confirmar que el ANN no se está equivocando.

## Errores comunes

Activar un índice ANN en producción sin haber medido antes, contra el k-NN exacto, cuánto recall estás perdiendo.

## Cómo crece el costo

El siguiente gráfico muestra por qué, a partir de cierto tamaño de corpus, comparar contra todos deja de ser viable.

```python
import matplotlib.pyplot as plt

cantidad_faqs = [100, 1_000, 10_000, 100_000]
trabajo = [n / 100 for n in cantidad_faqs]

plt.plot(cantidad_faqs, trabajo, marker="o")
plt.xscale("log")
plt.xlabel("cantidad de FAQs")
plt.ylabel("trabajo relativo por pregunta")
plt.title("k-NN exacto: el costo crece con el tamano del corpus")
plt.grid(alpha=.3)
plt.show()
```
