# Recall@K para ANN

## Problema real

Velocidad con evidencia de calidad

## Conceptos clave

ANN reduce búsqueda, pero puede omitir vecinos. Recall@K compara sus resultados con el top exacto en las mismas consultas.

```python
# Ejemplo ejecutable: Recall@K para ANN
exact={"a","b","c","d","e"};ann={"a","b","e","f","g"}
print("Recall@5:",len(exact&ann)/len(exact))
```

## Visualización

El gráfico siguiente resume el concepto técnico presentado.


```python
# Ejemplo ejecutable: Recall@K para ANN
import matplotlib.pyplot as plt
settings=[1,4,8,16];recall=[.61,.79,.88,.94];latency=[2,3,5,9]
fig,ax=plt.subplots(figsize=(6,4));ax.plot(settings,recall,"o-",label="Recall@5");ax.set_ylim(0,1);ax.set_xlabel("esfuerzo de búsqueda")
ax2=ax.twinx();ax2.plot(settings,latency,"s-",color="peru",label="latencia");ax2.set_ylabel("ms");ax.legend(loc="lower right");plt.title("Trade-off ANN");plt.show()
```
