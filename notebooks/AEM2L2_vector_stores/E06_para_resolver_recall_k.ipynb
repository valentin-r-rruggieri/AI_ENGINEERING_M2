# Recall@K: ¿tu búsqueda encuentra lo que importa?

## Problema real

Un índice rápido que no encuentra la FAQ correcta no sirve

## Conceptos clave

Recall@K responde una pregunta simple: de los chunks que el k-NN exacto considera correctos, ¿cuántos aparecieron también en el Top-K que devolvió tu índice ANN? Si Recall@5 es 0.6, tu chatbot se está perdiendo 4 de cada 10 FAQs relevantes.

```python
# "correctos" = lo que devolveria la busqueda exacta. "encontrados" = lo que devolvio el indice ANN.
correctos = {"a", "b", "c", "d", "e"}
encontrados = {"a", "b", "e", "f", "g"}

recall_at_5 = len(correctos & encontrados) / len(correctos)
print("Recall@5:", recall_at_5, "-> SoporteBot encontro", len(correctos & encontrados), "de", len(correctos), "FAQs relevantes")
```

## Cómo usar este número

Recall@K no se mide una vez y listo: lo volvés a medir cada vez que cambiás un parámetro del índice (por ejemplo `nprobe` en IVF o `efSearch` en HNSW). Así podés detectar si al ganar velocidad, sin darte cuenta, bajó la calidad de las respuestas.

## Errores comunes

Cambiar la configuración del índice para ganar velocidad y no volver a medir Recall@K: el chatbot puede empezar a fallar en preguntas que antes contestaba bien, sin que nadie lo note hasta que un usuario se queja.

## El trade-off entre velocidad y calidad

Cuanto más "esfuerzo de búsqueda" le pedís al índice ANN, más sube el recall -pero también la latencia. El siguiente gráfico muestra ese balance.

```python
import matplotlib.pyplot as plt

esfuerzo_busqueda = [1, 4, 8, 16]
recall = [.61, .79, .88, .94]
latencia_ms = [2, 3, 5, 9]

fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(esfuerzo_busqueda, recall, "o-", label="Recall@5")
ax.set_ylim(0, 1)
ax.set_xlabel("esfuerzo de busqueda del indice (ej: nprobe, efSearch)")
ax.set_ylabel("Recall@5")

ax2 = ax.twinx()
ax2.plot(esfuerzo_busqueda, latencia_ms, "s-", color="peru", label="latencia")
ax2.set_ylabel("latencia (ms)")

ax.legend(loc="lower right")
plt.title("Mas esfuerzo de busqueda = mejor recall, pero mas lento")
plt.show()
```
