# La unidad de un vector store

## Problema real

De un chunk de texto a un registro que se puede recuperar y explicar

## Conceptos clave

Un vector store no es solo una lista de números. Cada fila conecta un ID, el embedding, el texto original y la metadata necesaria para poder mostrarle la respuesta —y la fuente— al usuario de SoporteBot.

```python
# Asi luce un registro real dentro del vector store de SoporteBot.
registro = {
    "chunk_id": "rrhh-007",
    "embedding_dim": 1536,
    "content": "Las vacaciones requieren aprobacion del manager.",
    "metadata": {"source": "faq_rrhh.md", "department": "rrhh", "version": "v1"},
}

for campo, valor in registro.items():
    print(f"{campo}: {valor}")
```

## Por qué cada campo importa

- `chunk_id`: te permite rastrear de dónde salió una respuesta si un usuario se queja de que está mal.
- `embedding_dim`: tiene que coincidir con lo que espera tu índice; si no, el vector store lo rechaza.
- `content`: es lo que realmente lee el LLM para responder.
- `metadata`: es lo que te deja filtrar (por ejemplo, "solo FAQs de RRHH") antes o después de buscar.

## Errores comunes

Guardar el vector sin `content` ni `metadata`: la búsqueda va a funcionar, pero no vas a poder explicarle al usuario -ni a vos mismo- por qué el chatbot devolvió ese resultado.

## Las 4 piezas de un registro

El gráfico siguiente resume las cuatro piezas que siempre debería tener un registro de tu vector store.

```python
import matplotlib.pyplot as plt

partes = ["chunk_id", "embedding", "content", "metadata"]
plt.figure(figsize=(7, 3))
for i, parte in enumerate(partes):
    plt.barh([i], [1], left=[0], color=["#264653", "#2a9d8f", "#e9c46a", "#f4a261"][i])
    plt.text(.5, i, parte, ha="center", va="center", color="white" if i < 2 else "black")
plt.xlim(0, 1)
plt.yticks([])
plt.title("Un registro de vector store es trazable: podes explicar cada resultado")
plt.show()
```
