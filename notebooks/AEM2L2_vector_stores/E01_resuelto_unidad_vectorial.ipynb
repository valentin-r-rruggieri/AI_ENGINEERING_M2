# La unidad de un vector store

## Problema real

Del chunk a un registro recuperable

## Conceptos clave

Un vector store no guarda solo números: conecta un ID, un embedding, el texto y metadata que permite recuperar y filtrar evidencia.

```python
# Ejemplo ejecutable: La unidad de un vector store
record={"chunk_id":"faq-007","embedding_dim":1536,"content":"Las vacaciones requieren aprobación.","metadata":{"source":"faq.txt","department":"rrhh","version":"v1"}}
for key,value in record.items(): print(f"{key}: {value}")
```

## Visualización

El gráfico siguiente resume el concepto técnico presentado.


```python
# Ejemplo ejecutable: La unidad de un vector store
import matplotlib.pyplot as plt
layers=["chunk_id","embedding","contenido","metadata"]
plt.figure(figsize=(7,3))
for i,label in enumerate(layers):
    plt.barh([i],[1],left=[0],color=["#264653","#2a9d8f","#e9c46a","#f4a261"][i])
    plt.text(.5,i,label,ha="center",va="center",color="white" if i<2 else "black")
plt.xlim(0,1);plt.yticks([]);plt.title("Un registro vectorial es una unidad trazable");plt.show()
```
