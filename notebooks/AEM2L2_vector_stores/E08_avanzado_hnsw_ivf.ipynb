# HNSW e IVF: dos formas de no comparar contra todos

## Problema real

Cuando el corpus crece, ni el k-NN exacto ni FAISS Flat alcanzan

## Conceptos clave

**HNSW** arma un grafo donde cada vector está conectado a sus vecinos más cercanos, y navega ese grafo en vez de mirar todo el corpus. **IVF** agrupa los vectores en "listas" (clusters) y, en cada búsqueda, sólo revisa las listas más prometedoras. Los dos te dan velocidad a cambio de un poco de recall -y cada uno tiene su propio parámetro para ajustar ese balance.

```python
# Que parametro ajustar segun el tipo de indice que use tu vector store.
indices = {
    "HNSW": {"usa": "grafo de vecinos", "memoria": "alta", "parametro_para_ajustar": "efSearch"},
    "IVF": {"usa": "listas (clusters)", "memoria": "media", "parametro_para_ajustar": "nprobe"},
}

for nombre, config in indices.items():
    print(nombre, config)
```

## Cuál elegir

Si tenés memoria disponible y querés el mejor recall posible a alta velocidad, HNSW suele ganar. Si el corpus es enorme y la memoria es limitada, IVF (o una combinación IVF+HNSW) es más habitual. En los dos casos, la única forma de saber si la elección fue buena es medir Recall@K en tus propias FAQs, como en el ejercicio anterior.

## Errores comunes

Copiar un valor de `nprobe` o `efSearch` de un tutorial de internet sin probarlo contra tu propio corpus: el balance ideal depende de cuántos chunks tenés y de qué tan parecidos son entre sí.

## Cómo se ve cada estrategia

A la izquierda, HNSW navega saltando entre vecinos cercanos. A la derecha, IVF descarta listas enteras de vectores sin revisarlas.

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(8, 3))
axes[0].plot([0, 1, 2, 1.5, 2.5], [0, 1, .2, 2, 1.5], "o-")
axes[0].set_title("HNSW: navegar vecinos")
axes[1].bar(["lista 1", "lista 2", "lista 3"], [8, 2, 7], color=["#ccc", "#f4a261", "#ccc"])
axes[1].set_title("IVF: explorar solo las listas prometedoras")
for ax in axes:
    ax.grid(alpha=.2)
plt.show()
```
