# Filtros de metadata

## Problema real

La similitud semántica no sabe de permisos

## Conceptos clave

Que un chunk sea semánticamente parecido a la pregunta no significa que el usuario deba verlo. Si SoporteBot filtra por departamento, primero buscás por similitud y **después** aplicás las reglas de acceso: la metadata es lo que te permite hacer ese filtro.

```python
# El vector store ya devolvio los candidatos mas parecidos a la pregunta.
# Ahora nos quedamos solo con los que puede ver un usuario de RRHH.
resultados = [
    {"id": "a", "score": .92, "department": "rrhh"},
    {"id": "b", "score": .89, "department": "finanzas"},
    {"id": "c", "score": .81, "department": "rrhh"},
]

visibles_para_rrhh = [fila for fila in resultados if fila["department"] == "rrhh"]
print(visibles_para_rrhh)
```

## Por qué filtrar por metadata, no por texto

Filtrar por `department` es rápido y confiable porque es un campo estructurado. Intentar "adivinar" el departamento leyendo el texto del chunk es lento y poco confiable -por eso la metadata se define al momento de indexar, no al momento de responder.

## Errores comunes

Armar el prompt para el LLM con los resultados y filtrar recién ahí: para ese momento, un chunk que el usuario no debía ver ya viajó por tu backend (y puede haber quedado en un log).

## Cuántos chunks pasaron el filtro

El siguiente gráfico muestra cuántos chunks de cada departamento pasaron un filtro de metadata, sobre el total recuperado por similitud.

```python
import matplotlib.pyplot as plt

labels = ["RRHH", "Finanzas", "Tecnología"]
values = [5, 3, 4]
plt.bar(labels, values, color=["#2a9d8f", "#e76f51", "#457b9d"])
plt.title("Candidatos por departamento, despues del filtro de metadata")
plt.ylabel("chunks")
plt.show()
```
