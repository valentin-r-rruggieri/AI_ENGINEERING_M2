# Filtros de metadata

## Problema real

Relevancia con restricciones

## Conceptos clave

La similitud semántica no reemplaza reglas de autorización o dominio. Recuperá candidatos y aplicá filtros coherentes con la política de acceso.

```python
# Ejemplo ejecutable: Filtros de metadata
results=[{"id":"a","score":.92,"department":"rrhh"},{"id":"b","score":.89,"department":"finanzas"},{"id":"c","score":.81,"department":"rrhh"}]
print([row for row in results if row["department"]=="rrhh"])
```
