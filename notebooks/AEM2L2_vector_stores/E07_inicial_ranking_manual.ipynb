# Auditar un ranking a mano

## Problema real

Un score sin contexto no te dice si la respuesta es buena

## Conceptos clave

Antes de conectar el retrieval al LLM, conviene mirar el ranking a mano: ¿el score más alto corresponde a la FAQ correcta? Auditar esto, aunque sea con pocos ejemplos, te ahorra debuggear un chatbot que "alucina" cuando en realidad el problema es que le llegó el chunk equivocado.

```python
# Ranking de FAQs de SoporteBot para la pregunta "¿como pido vacaciones?"
resultados = [("faq-licencias", .87), ("faq-vacaciones", .78), ("faq-facturacion", .22)]

for chunk_id, score in sorted(resultados, key=lambda fila: fila[1], reverse=True):
    print(chunk_id, score)
```

## Qué mirar al auditar

Que el orden tenga sentido (tiene lógica que "facturación" aparezca tan abajo para una pregunta de vacaciones, no es relevante), que no haya empates raros, y que el score más alto sea, a simple vista, el chunk que vos elegirías como humano.

## Errores comunes

Ignorar empates o valores fuera de rango (por ejemplo, un score de coseno de 1.4, que es imposible) solo porque "el número más alto parece razonable".

## El ranking, de un vistazo

El siguiente gráfico muestra el mismo ranking de forma visual: útil para detectar de un vistazo si un resultado quedó mal ubicado.

```python
import matplotlib.pyplot as plt

scores = {"licencias": .87, "vacaciones": .78, "facturación": .22}
plt.barh(list(scores), list(scores.values()), color="#457b9d")
plt.xlim(0, 1)
plt.title("Auditoría de ranking: ¿coincide con lo que esperarías?")
plt.show()
```
