## Problema real

## Conceptos clave

Un ranking correcto indica métrica, normalización, orden y contenido asociado. Mirar solo un número no alcanza para validar relevancia.

```python
# Ejemplo ejecutable: auditar un ranking
rows=[("faq-2",.87),("faq-7",.78),("faq-9",.22)]
for item in sorted(rows,key=lambda x:x[1],reverse=True):print(item)
```
