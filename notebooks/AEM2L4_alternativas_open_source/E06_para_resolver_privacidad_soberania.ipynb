## Conceptos clave

```python
# Ejemplo ejecutable: Privacidad y soberanía de datos
import matplotlib.pyplot as plt
plt.scatter([1,2,3],[3,2,1],s=[300,300,300],c=["#e76f51","#e9c46a","#2a9d8f"]);plt.text(1,3,"PII",ha="center");plt.text(2,2,"interno",ha="center");plt.text(3,1,"público",ha="center");plt.title("Clasificación de sensibilidad");plt.show()
```
