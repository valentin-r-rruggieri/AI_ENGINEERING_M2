# Benchmark reproducible

## Problema real

La evidencia debe poder repetirse

## Conceptos clave

El benchmark fija corpus, queries, métricas, hardware, versión de modelo y configuración. Sin ese registro no existe comparación defendible.

```python
# Ejemplo ejecutable: Benchmark reproducible
import matplotlib.pyplot as plt
runs=[1,2,3,4];plt.plot(runs,[.79,.81,.80,.82],"o-",label="Recall");plt.plot(runs,[7,6,8,7],"s-",label="p95 ms");plt.legend();plt.title("Repeticiones del benchmark");plt.show()
```
