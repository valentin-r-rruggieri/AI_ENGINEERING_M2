## Modelo mental

Una ADR concisa explica contexto, alternativas, evidencia, decisión, consecuencias y fecha de revisión.

```python
import matplotlib.pyplot as plt
blocks=["contexto","alternativas","evidencia","decisión","revisión"]
plt.figure(figsize=(8,2));plt.plot(range(5),[0]*5,"o-",color="#6a4c93")
for i,b in enumerate(blocks):plt.text(i,.1,b,ha="center");plt.axis("off")
plt.title("ADR breve y auditable");plt.show()
```
