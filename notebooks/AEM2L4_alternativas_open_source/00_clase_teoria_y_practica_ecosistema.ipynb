## De qué se trata

La elección entre API administrada, modelos locales o una arquitectura híbrida no se resuelve por moda. Depende de calidad, costo, latencia, privacidad, hardware y capacidad operativa.

## 1. API o modelo local

| Criterio | API administrada | Modelo local |
|---|---|---|
| Inicio | Rápido | Requiere instalación |
| Operación | Proveedor | Equipo propio |
| Datos | Viajan al proveedor según configuración | Permanecen en el entorno |
| Escala | Administrada | Depende de infraestructura |
| Costo | Por uso | Hardware + energía + mantenimiento |
| Control | Menor | Mayor |

```python
# Ejemplo ejecutable: Clase 4 — Alternativas y ecosistema open-source
import matplotlib.pyplot as plt

criteria = ["calidad", "privacidad", "latencia", "operación"]
api = [4, 2, 4, 5]
local = [3, 5, 3, 2]
plt.plot(criteria, api, "o-", label="API administrada")
plt.plot(criteria, local, "s-", label="local")
plt.ylim(0, 5)
plt.ylabel("puntaje ilustrativo")
plt.title("Una decisión requiere varios criterios")
plt.legend()
plt.grid(alpha=.25)
plt.show()
```

## 2. Calidad, benchmark y cambios seguros

Para comparar opciones mantené fijo: corpus, golden cases, chunking, métrica, Top-K y hardware/red documentados. Medí Recall@K, p50/p95, throughput, costo estimado y fallos.

Shadow traffic envía una copia de consultas a una alternativa nueva sin alterar la respuesta real. Un circuit breaker permite fallar hacia un camino seguro cuando un proveedor se degrada.

## Tabla de decisión auditable

| Decisión | Evidencia necesaria | Riesgo | Mitigación |
|---|---|---|---|
| API | Golden set + costo real | rate limit/proveedor | caché y fallback |
| Local | benchmark + hardware | operación/modelo | monitoreo y versionado |
| Híbrida | comparación continua | complejidad | router, shadow y métricas |

```python
# Ejemplo ejecutable: Clase 4 — Alternativas y ecosistema open-source
import matplotlib.pyplot as plt

backends = ["API", "local"]
recall = [.86, .83]
p95 = [420, 180]
fig, ax = plt.subplots(figsize=(6,4))
ax.bar(backends, recall, color="#2a9d8f")
ax.set_ylim(0,1); ax.set_ylabel("Recall@5")
ax2 = ax.twinx()
ax2.plot(backends, p95, "o-", color="#e76f51")
ax2.set_ylabel("p95 ms")
plt.title("Calidad y latencia deben leerse juntas")
plt.show()
```

## Ejemplo ejecutable

El bloque siguiente ilustra una implementación mínima del concepto.


```python
# Ejemplo ejecutable: Clase 4 — Alternativas y ecosistema open-source
weights = {"quality": .30, "privacy": .30, "latency": .20, "operations": .20}
api = {"quality": 4, "privacy": 2, "latency": 4, "operations": 5}
local = {"quality": 3, "privacy": 5, "latency": 3, "operations": 2}

def weighted_score(option):
    return sum(weights[key] * option[key] for key in weights)

print("API:", round(weighted_score(api), 2))
print("Local:", round(weighted_score(local), 2))
print("Cambiá los pesos y explicá qué necesidad del negocio representa el cambio.")
```
