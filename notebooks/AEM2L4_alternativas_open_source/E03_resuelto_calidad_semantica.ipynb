# Calidad semántica con golden cases

## Problema real

La dimensión no garantiza relevancia

## Conceptos clave

La calidad se mide con preguntas y evidencias reales. Un modelo más grande puede mejorar ambigüedad, pero no corrige chunks defectuosos.

```python
# Ejemplo ejecutable: Calidad semántica con golden cases
import matplotlib.pyplot as plt
models=["A","B","C"];recall=[.72,.84,.81];plt.bar(models,recall,color="#2a9d8f");plt.ylim(0,1);plt.title("Golden cases por modelo");plt.show()
```

## Ejemplo ejecutable

El bloque siguiente ilustra una implementación mínima del concepto.


```python
# Ejemplo ejecutable: Calidad semántica con golden cases
golden={"modelo_A":[1,0,1],"modelo_B":[1,1,1]};print({k:sum(v)/len(v) for k,v in golden.items()})
```
