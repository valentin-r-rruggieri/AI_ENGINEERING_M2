## Conceptos clave

Una decisión responsable incorpora volumen, storage, red, hardware, mantenimiento y tiempo del equipo.

```python
# Ejemplo ejecutable: Estimar costo y operación
import matplotlib.pyplot as plt
items=["tokens","storage","hardware","operación"];values=[.2,.1,.4,.3];plt.pie(values,labels=items,autopct="%1.0f%%");plt.title("Costo total ilustrativo");plt.show()
```
