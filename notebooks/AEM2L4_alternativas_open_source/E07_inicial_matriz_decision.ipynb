## Modelo mental

Una matriz no reemplaza el juicio, pero vuelve visibles pesos, supuestos y desacuerdos entre calidad, costo, privacidad y operación.

```python
import matplotlib.pyplot as plt
criteria=["calidad","privacidad","latencia","operación"];api=[4,2,4,5];local=[3,5,3,2]
plt.plot(criteria,api,"o-",label="API");plt.plot(criteria,local,"o-",label="local");plt.ylim(0,5);plt.legend();plt.title("Matriz de decisión");plt.show()
```
