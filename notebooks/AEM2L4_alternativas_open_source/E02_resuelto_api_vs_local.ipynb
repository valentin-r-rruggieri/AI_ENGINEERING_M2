## Conceptos clave

La API abstrae infraestructura; la ejecución local controla datos y operación. La calidad debe evaluarse con el mismo set de consultas.

```python
# Ejemplo ejecutable: API administrada frente a inferencia local
import matplotlib.pyplot as plt
for y,label in enumerate(["app","API externa","vector"]):plt.text(y,0,label,ha="center",bbox={"boxstyle":"round","fc":"#dbe9f6"})
plt.plot([0,1,2],[0,0,0],"->");plt.xlim(-.5,2.5);plt.ylim(-1,1);plt.axis("off");plt.title("Flujo administrado");plt.show()
```
