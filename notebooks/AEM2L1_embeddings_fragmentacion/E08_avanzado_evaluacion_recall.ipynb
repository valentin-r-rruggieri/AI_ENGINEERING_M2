## Modelo mental

Recall@K responde si el sistema recuperó al menos una evidencia esperada. Es una métrica de retrieval, no de calidad lingüística de la respuesta.

**Pregunta guía:** ¿qué decisión de ingeniería cambia el resultado y cómo la medirías?

```python
import matplotlib.pyplot as plt
k=[1,2,3,5]; recall=[.58,.76,.89,.91]; tokens=[180,320,470,760]
fig, ax=plt.subplots(figsize=(6,4)); ax.plot(k,recall,marker="o"); ax.set(xlabel="K",ylabel="Recall@K",ylim=(0,1),title="Calidad vs. presupuesto")
ax2=ax.twinx(); ax2.plot(k,tokens,color="peru",marker="s"); ax2.set_ylabel("tokens de contexto"); plt.show()
```

## Práctica guiada

El gráfico separa calidad de costo. Elegí un K que tenga evidencia suficiente sin llenar el prompt de ruido.

```python
golden=[{"question":"¿Cómo recupero mi contraseña?","expected":"faq-003","retrieved":["faq-003","faq-008"]},{"question":"¿Cómo solicito vacaciones?","expected":"faq-007","retrieved":["faq-002","faq-007"]}]
recall=sum(case["expected"] in case["retrieved"] for case in golden)/len(golden)
print("Recall@2 =", recall)
```
