# Embeddings API con trazabilidad

## Problema real

Persistir configuración y dimensiones

## Conceptos clave

La API reduce infraestructura, pero obliga a registrar modelo, dimensión, fecha y parámetros de chunking para poder reconstruir un índice.

```python
# Ejemplo ejecutable: Embeddings API con trazabilidad
import json, datetime
manifest={"embedding_model":"text-embedding-3-small","chunk_size":160,"overlap":24,"created_at":datetime.date.today().isoformat(),"dimension":None}
print(json.dumps(manifest,ensure_ascii=False,indent=2))
```

## Ejemplo ejecutable

El bloque siguiente ilustra una implementación mínima del concepto.


```python
# Ejemplo ejecutable: Embeddings API con trazabilidad
import os
if os.getenv("OPENAI_API_KEY"):
    from openai import OpenAI
    response=OpenAI().embeddings.create(model=os.getenv("AEM2_EMBEDDING_MODEL","text-embedding-3-small"),input=["política de vacaciones"])
    manifest["dimension"]=len(response.data[0].embedding)
    print(manifest)
else:
    print("API no configurada: el manifiesto sigue siendo útil para documentar el experimento.")
```
