## Práctica guiada

La primera parte crea el manifiesto sin red; la segunda invoca la API únicamente si existe una clave.

```python
import os
if os.getenv("OPENAI_API_KEY"):
    from openai import OpenAI
    response=OpenAI().embeddings.create(model=os.getenv("AEM2_EMBEDDING_MODEL","text-embedding-3-small"),input=["política de vacaciones"])
    manifest["dimension"]=len(response.data[0].embedding)
    print(manifest)
else:
    print("API no configurada: el manifiesto sigue siendo útil para documentar el experimento.")
```
