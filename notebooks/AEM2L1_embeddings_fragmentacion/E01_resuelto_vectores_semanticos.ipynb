## Modelo mental

Representar textos por su significado permite acercar 'restablecer credenciales' y 'reiniciar contraseña' aunque no compartan el mismo vocabulario.

**Pregunta guía:** ¿qué decisión de ingeniería cambia el resultado y cómo la medirías?

```python
import matplotlib.pyplot as plt
points = {"contraseña": (0.8, 1.0), "credenciales": (1.2, 1.3), "vacaciones": (4.4, 3.8)}
fig, ax = plt.subplots(figsize=(6, 4))
for label, (x, y) in points.items():
    ax.scatter(x, y, s=110)
    ax.annotate(label, (x + .08, y + .08))
ax.set(title="Analogía: proximidad semántica", xlim=(0, 5), ylim=(0, 5), xlabel="dimensión 1", ylabel="dimensión 2")
ax.grid(alpha=.25)
plt.show()
```
