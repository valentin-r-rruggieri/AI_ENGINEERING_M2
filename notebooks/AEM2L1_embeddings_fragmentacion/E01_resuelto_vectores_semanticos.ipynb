# Vectores y significado semántico

## Problema real

Por qué una búsqueda por palabras falla

## Conceptos clave

Representar textos por su significado permite acercar 'restablecer credenciales' y 'reiniciar contraseña' aunque no compartan el mismo vocabulario.

```python
# Ejemplo ejecutable: Vectores y significado semántico
import matplotlib.pyplot as plt
points = {"contraseña": (0.8, 1.0), "credenciales": (1.2, 1.3), "vacaciones": (4.4, 3.8)}
fig, ax = plt.subplots(figsize=(6, 4))
for label, (x, y) in points.items():
    ax.scatter(x, y, s=110)
    ax.annotate(label, (x + .08, y + .08))
ax.set(title="Analogía: proximidad semántica", xlim=(0, 5), ylim=(0, 5), xlabel="dimensión 1", ylabel="dimensión 2")
ax.grid(alpha=.25)
plt.show()
```

## Ejemplo ejecutable

El bloque siguiente ilustra una implementación mínima del concepto.


```python
# Ejemplo ejecutable: Vectores y significado semántico
from math import dist
print("distancia contraseña-credenciales:", round(dist((.8,1.0),(1.2,1.3)), 2))
print("distancia contraseña-vacaciones:", round(dist((.8,1.0),(4.4,3.8)), 2))
```
