# Elegir chunk_size con evidencia

## Problema real

No existe un tamaño universal

## Conceptos clave

El tamaño del chunk es una hipótesis. Debe contrastarse con consultas reales, recall, costo y legibilidad del contexto recuperado.

```python
# Ejemplo ejecutable: Elegir chunk_size con evidencia
import matplotlib.pyplot as plt
sizes, recall, context = [80,160,320], [.72,.86,.81], [1.2,2.1,4.0]
fig, ax = plt.subplots(figsize=(6,4))
ax.plot(sizes, recall, marker="o", label="Recall@K")
ax.set(xlabel="palabras por chunk", ylabel="Recall@K", ylim=(0,1), title="Hipótesis de chunk_size")
ax2=ax.twinx(); ax2.plot(sizes, context, marker="s", color="peru", label="contexto")
ax2.set_ylabel("contexto enviado (unidades relativas)"); plt.show()
```

## Ejemplo ejecutable

El bloque siguiente ilustra una implementación mínima del concepto.


```python
# Ejemplo ejecutable: Elegir chunk_size con evidencia
experiments = [{"size":80,"overlap":12,"recall":.72},{"size":160,"overlap":24,"recall":.86},{"size":320,"overlap":48,"recall":.81}]
best=max(experiments,key=lambda row:row["recall"])
print("Configuración candidata:", best)
```
