# Overlap y procedencia

## Problema real

Evitar que la evidencia se pierda en los bordes

## Conceptos clave

El overlap repite una franja controlada. La metadata permite explicar de qué documento y posición salió cada chunk recuperado.

```python
# Ejemplo ejecutable: Overlap y procedencia
import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize=(9,2.2))
ax.broken_barh([(0,7),(4,7)], (4,4), facecolors=["#9ecae1","#fdae6b"])
ax.text(1,6,"chunk 1",ha="center"); ax.text(9,6,"chunk 2",ha="center")
ax.axvspan(4,7,color="#d9ef8b",alpha=.7); ax.text(5.5,3,"overlap",ha="center")
ax.set(xlim=(0,12), ylim=(0,10), yticks=[], xlabel="posición en el documento", title="La zona verde aparece en ambos chunks")
plt.show()
```

## Ejemplo ejecutable

El bloque siguiente ilustra una implementación mínima del concepto.


```python
# Ejemplo ejecutable: Overlap y procedencia
def chunks_with_overlap(words, size=7, overlap=3):
    return [" ".join(words[i:i+size]) for i in range(0, len(words), size-overlap)]
sample = "La política de vacaciones requiere aprobación del líder antes de confirmar".split()
print(chunks_with_overlap(sample))
```

## Consideraciones técnicas

Usar overlap enorme por defecto; omitir source, chunk_id o posición de origen.
