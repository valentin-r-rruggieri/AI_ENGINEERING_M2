## Modelo mental

El coseno compara el ángulo entre vectores; el producto punto incorpora además la longitud. La normalización cambia la interpretación del ranking.

**Pregunta guía:** ¿qué decisión de ingeniería cambia el resultado y cómo la medirías?

```python
import numpy as np, matplotlib.pyplot as plt
a, b = np.array([3., 2.]), np.array([1.5, 1.])
fig, ax = plt.subplots(figsize=(5, 4))
for vector, color, label in [(a, "navy", "consulta"), (b, "peru", "documento")]:
    ax.quiver(0, 0, *vector, angles="xy", scale_units="xy", scale=1, color=color, label=label)
ax.set(xlim=(0,4), ylim=(0,3), aspect="equal", title="Misma dirección, distinta magnitud")
ax.grid(alpha=.3); ax.legend(); plt.show()
```

## Antes de ejecutar

Predicción: si dos vectores apuntan igual pero uno es más largo, ¿qué métrica cambia más?

Escribí una hipótesis breve. Luego contrastala con el resultado visual y numérico.

## Práctica guiada

Compará el valor numérico con el dibujo: primero hacé la predicción, después explicá el resultado.

```python
cosine = np.dot(a,b)/(np.linalg.norm(a)*np.linalg.norm(b))
dot = np.dot(a,b)
print({"coseno": round(float(cosine),3), "producto_punto": round(float(dot),3)})
```
