# Chunking con sentido

## Problema real

Un documento entero diluye la evidencia

## Conceptos clave

Fragmentar no es cortar al azar: cada chunk debe poder responder una parte coherente de una consulta y conservar contexto suficiente.

```python
# Ejemplo ejecutable: Chunking con sentido
import matplotlib.pyplot as plt
words = "Para recuperar la contraseña elegí olvidé mi contraseña y revisá tu correo laboral".split()
fig, ax = plt.subplots(figsize=(9,2))
for i, word in enumerate(words):
    ax.text(i, 0, word, ha="center", va="center", bbox={"boxstyle":"round","fc":"#e8eef7"})
ax.axvspan(-.5, 4.5, color="#f0b27a", alpha=.35, label="chunk 1")
ax.axvspan(4.5, len(words)-.5, color="#85c1e9", alpha=.35, label="chunk 2")
ax.set(xlim=(-1,len(words)), ylim=(-1,1), yticks=[], title="Un corte debe preservar una unidad de sentido"); ax.legend(); plt.show()
```

## Ejemplo ejecutable

El bloque siguiente ilustra una implementación mínima del concepto.


```python
# Ejemplo ejecutable: Chunking con sentido
def split_words(text, size):
    words = text.split()
    return [" ".join(words[i:i+size]) for i in range(0, len(words), size)]
print(split_words(" ".join(words), 5))
```
