**Pregunta guía:** ¿qué decisión de ingeniería cambia el resultado y cómo la medirías?

```python
import matplotlib.pyplot as plt
steps=["documento","chunks","embeddings","Top-K","contexto"]
plt.figure(figsize=(9,2.2)); plt.plot(range(len(steps)),[1]*len(steps),"o-",color="#1f4e79")
for i,label in enumerate(steps): plt.text(i,1.08,label,ha="center")
plt.axis("off"); plt.title("Pipeline de recuperación trazable"); plt.show()
```
