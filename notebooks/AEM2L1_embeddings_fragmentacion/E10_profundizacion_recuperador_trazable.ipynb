# recuperador trazable

## Problema real

De texto a evidencia para RAG

## Conceptos clave

Un recuperador útil entrega contenido, score, ID y procedencia. El LLM recibe evidencia acotada, no el documento completo.

```python
# Ejemplo ejecutable: recuperador trazable
import matplotlib.pyplot as plt
steps=["documento","chunks","embeddings","Top-K","contexto"]
plt.figure(figsize=(9,2.2)); plt.plot(range(len(steps)),[1]*len(steps),"o-",color="#1f4e79")
for i,label in enumerate(steps): plt.text(i,1.08,label,ha="center")
plt.axis("off"); plt.title("Pipeline de recuperación trazable"); plt.show()
```

## Ejemplo ejecutable

El bloque siguiente ilustra una implementación mínima del concepto.


```python
# Ejemplo ejecutable: recuperador trazable
results=[{"chunk_id":"faq-003","score":.91,"text":"El enlace de contraseña vence en 30 minutos.","source":"faq.txt"},{"chunk_id":"faq-008","score":.79,"text":"Soporte verifica el correo laboral.","source":"faq.txt"}]
context="\n\n".join(f"[{r['chunk_id']} | {r['source']} | {r['score']:.2f}]\n{r['text']}" for r in results)
print(context)
```
