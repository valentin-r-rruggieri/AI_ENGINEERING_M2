# Embeddings reales y Top-K

## Problema real

Mismo modelo para documentos y consulta

## Conceptos clave

Un índice solo es coherente si corpus y consulta comparten el mismo espacio vectorial. La celda opcional usa OpenAI; el ranking de prueba funciona sin clave.

```python
# Ejemplo ejecutable: Embeddings reales y Top-K
import matplotlib.pyplot as plt
scores={"faq-003":.91,"faq-008":.79,"faq-002":.41,"faq-011":.24}
plt.bar(scores.keys(),scores.values(),color=["#1f77b4","#1f77b4","#b0b0b0","#b0b0b0"])
plt.axhline(.5,color="peru",linestyle="--",label="threshold ilustrativo")
plt.ylim(0,1); plt.title("Top-K debe conservar score e ID"); plt.legend(); plt.show()
```

## Ejemplo ejecutable

El bloque siguiente ilustra una implementación mínima del concepto.


```python
# Ejemplo ejecutable: Embeddings reales y Top-K
import os
if not os.getenv("OPENAI_API_KEY"):
    print("Modo sin API: configurá OPENAI_API_KEY para generar embeddings reales.")
else:
    from openai import OpenAI
    client=OpenAI()
    result=client.embeddings.create(model=os.getenv("AEM2_EMBEDDING_MODEL","text-embedding-3-small"), input=["recuperar contraseña"])
    print("dimensión:", len(result.data[0].embedding))
```

## Consideraciones técnicas

Mezclar modelos; perder el orden de los scores; no conservar los IDs de fuente.
