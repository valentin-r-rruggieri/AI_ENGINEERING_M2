## Conceptos clave

Antes de usar una base vectorial, conviene poder inspeccionar un ranking pequeño. Eso permite detectar una métrica invertida o una normalización inconsistente.

```python
# Ejemplo ejecutable: ranking manual
import numpy as np, matplotlib.pyplot as plt
labels=["contraseña","vacaciones","facturación"]
scores=[.93,.33,.18]
plt.barh(labels,scores,color="#4c78a8"); plt.xlim(0,1); plt.title("Ranking manual de una consulta"); plt.show()
```

## Ejemplo ejecutable

El bloque siguiente ilustra una implementación mínima del concepto.


```python
# Ejemplo ejecutable: ranking manual
query=np.array([1.,0.,1.])
docs={"contraseña":np.array([.9,0.,.8]),"vacaciones":np.array([0.,1.,.2])}
for name,vector in docs.items():
    score=np.dot(query,vector)/(np.linalg.norm(query)*np.linalg.norm(vector))
    print(name,round(float(score),3))
```

## Consideraciones técnicas

Tratar un score absoluto como universal; olvidar que cada modelo tiene una escala propia.
